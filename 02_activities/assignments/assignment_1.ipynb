```python
def anagram_checker(word_a, word_b):
    """ 
    This function checks if word_a and word_b contain the same letters
    word_a and word_b should be valid strings  
    
    """
    # Convert both words to lowercase, sort characters and compare them
    return sorted(word_a.lower()) == sorted(word_b.lower())
anagram_checker("Silent", "listen")
```

```python
anagram_checker("Silent", "Night")
```

```python
anagram_checker("night", "Thing")
```
